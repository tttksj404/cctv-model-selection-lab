# SOLIDER CCTV 속성 학습법 버블차트

x축은 Simuletic identity group-heldout 속성 정확도, y축은 PA-100K test mA, 버블 크기는 group exact rate이다. 200장 파일럿과 인증 다운로드 후 963장 부분집합을 함께 표시한다. 둘 다 실제 프로젝트 identity/track-heldout 증거가 아닌 속성 proxy이다.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt

path = Path('results/solider_simuletic_method_sweep_bubble_data.json')
data = json.loads(path.read_text(encoding='utf-8'))
rows = data['rows']
fig, ax = plt.subplots(figsize=(12, 7))
for row in rows:
    ax.scatter(row['simu_group_heldout_attribute_accuracy'], row['pa_test_mA'], s=900 * row['simu_group_exact_rate'], alpha=0.72, label=row['method'])
    ax.annotate(row['method'], (row['simu_group_heldout_attribute_accuracy'], row['pa_test_mA']), xytext=(6, 6), textcoords='offset points', fontsize=8)
ax.set_xlabel('Simuletic group-heldout attribute accuracy')
ax.set_ylabel('PA-100K test mA')
ax.set_title('SOLIDER CCTV Attribute Method Comparison')
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```
